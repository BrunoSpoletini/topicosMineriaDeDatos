## Métodos de Comparación de Resultados

```python
# Comentario: para comparar dos soluciones de clustering o una de ellas
#  contra las clases originales se suele usar una tabla, como por ejemplo:

def comparar_tabla(clusters1, clusters2):

    # Suponemos a los clusters numerados desde 0
    nclusters1 = np.max(clusters1) + 1
    nclusters2 = np.max(clusters2) + 1

    conf_matrix = np.zeros((nclusters1, nclusters2), dtype=int)

    for i in range(len(clusters1)):
        conf_matrix[clusters1[i], clusters2[i]] += 1

    print(conf_matrix)
    return conf_matrix


# pero se puede optimizar el match entre los dos clusterings, para hacer
# mejor la tabla, usando:

def comparar_tabla_optim(clusters1, clusters2, verbose=True):

    # Suponemos a los clusters numerados desde 0
    nclusters1 = np.max(clusters1) + 1
    nclusters2 = np.max(clusters2) + 1

    conf_matrix = np.zeros((nclusters1, nclusters2), dtype=int)

    for i in range(len(clusters1)):
        conf_matrix[clusters1[i], clusters2[i]] += 1

    # minimum weight matching, optimize the matching between a cluster
    # in clusters1 with one or more clusters in clusters2
    _, col_ind = linear_sum_assignment(-conf_matrix)
    optimized_conf_matrix = conf_matrix[:, col_ind]

    if verbose:
        print(optimized_conf_matrix)
    return optimized_conf_matrix

```

## Score de Estabilidad de Soluciones de Clustering

```python
# Código de ejemplo de como calcular el score de estabilidad de dos soluciones
# de clustering usando Iris

data = load_iris()
X = data.data
n = X.shape[0]

# Fijar el número de clusters
k = 5

# Crear dos muestras aleatorias del 90% de los datos
ind1 = np.random.choice(n, int(0.9 * n), replace=False)
ind2 = np.random.choice(n, int(0.9 * n), replace=False)

# Aplicar K-Means a ambas muestras
cc1 = KMeans(n_clusters=k, n_init=10, random_state=0).fit(X[ind1]).labels_
cc2 = KMeans(n_clusters=k, n_init=10, random_state=0).fit(X[ind2]).labels_

# Reinsertar los clusters en un vector de longitud n, con 0 en los puntos
# no seleccionados
v1 = np.zeros(n, dtype=int)
v2 = np.zeros(n, dtype=int)

# Se suman 5 a las etiquetas para permitir el truco de la raíz cuadrada
v1[ind1] = cc1 + 5
v2[ind2] = cc2 + 5

# Crear la matriz de similitud: 1 si están en el mismo cluster,
# -1 si están en distinto, 0 si alguno falta
def compute_similarity_matrix(v):
    a = np.sqrt(np.outer(v, v))
    m = np.divide(a, -a, out=np.zeros_like(a), where=a != 0) + 2 * (a == np.round(a))
    return m

def compute_similarity_matrix_fix(v):
    n = v.shape[0]
    is_valid = v > 0
    valid_mask = np.logical_and.outer(is_valid, is_valid)
    eq_mask = np.equal.outer(v, v)
    m = np.zeros((n,n), dtype=int)
    m[valid_mask & eq_mask] = 1
    m[valid_mask & ~eq_mask] = -1
    return m

m1 = compute_similarity_matrix_fix(v1)
m2 = compute_similarity_matrix_fix(v2)

# Calcular el score de estabilidad
validos = np.sum((v1 * v2) > 0)
if validos > 1:
    score = np.sum((m1 * m2)[np.triu_indices(n, k=1)] > 0) / (validos * (validos - 1) / 2)
else:
    score = 0  # Evita división por cero

print(score)
```

Es por esto que si uno de esos grupos se forma en la frontera, pueden tomar puntos de ambas clases, y arrastrar ese error hasta el final del proceso de clústering.

```python
# Gap Statistic: compara el log de la dispersión de los datos con el de B
# datasets uniformes en el rango de cada componente principal.
# Retorna el k elegido, el gap y el desvío s para cada k (el índice 0 es k=1)
def gap_statistic(X, k_max = 10, B = 20):
    # gereno datasets de referencia uniformes en el rango de cada componente principal
    X = PCA().fit_transform(X)
    X_ref = [np.random.uniform(X.min(axis=0), X.max(axis=0), size=X.shape) for _ in range(B)]

    gap = np.zeros(k_max)
    s = np.zeros(k_max)
    for k in range(1, k_max + 1):
        # la inercia de k-means es la dispersión W_k: suma de las distancias
        # al cuadrado de cada punto al centro de su cluster
        log_w = np.log(KMeans(n_clusters=k, n_init=10).fit(X).inertia_)
        log_w_ref = np.array([np.log(KMeans(n_clusters=k, n_init=10).fit(Xb).inertia_) for Xb in X_ref])
        gap[k - 1] = log_w_ref.mean() - log_w
        s[k - 1] = log_w_ref.std() * np.sqrt(1 + 1 / B)

    # elijo el menor k tal que gap(k) >= gap(k+1) - s(k+1)
    for k in range(1, k_max):
        if gap[k - 1] >= gap[k] - s[k]:
            return k, gap, s
    return k_max, gap, s
```

## Apartado b)

```python
# clustering k-means, retorna las etiquetas numeradas desde 0
def kmeans_labels(X, k):
    return KMeans(n_clusters=k, n_init=10).fit(X).labels_


# clustering jerárquico, retorna las etiquetas numeradas desde 0
def jer_labels(X, k):
    return fcluster(linkage(X, method="complete"), k, criterion="maxclust") - 1


# Estabilidad: para cada k hace clustering sobre m pares de submuestras con
# una fracción f de los datos, y mide qué tanto coinciden las dos soluciones
# en los puntos que comparten (1 = coinciden en todo).
# Retorna una matriz de (k_max - 1, m) con los scores, la fila i es k = i + 2
def estabilidad(X, k_max = 10, m = 20, f = 0.9, method=kmeans_labels):
    n = X.shape[0]
    scores = np.zeros((k_max - 1, m))

    for k in range(2, k_max + 1):
        for i in range(m):
            ind1 = np.random.choice(n, int(f * n), replace=False)
            ind2 = np.random.choice(n, int(f * n), replace=False)

            # reinserto los clusters en un vector de longitud n, sumo 1 para
            # que el 0 quede para los puntos que no se seleccionaron
            v1 = np.zeros(n, dtype=int)
            v2 = np.zeros(n, dtype=int)
            v1[ind1] = method(X[ind1], k) + 1
            v2[ind2] = method(X[ind2], k) + 1

            m1 = compute_similarity_matrix_fix(v1)
            m2 = compute_similarity_matrix_fix(v2)

            # proporción de pares (de los que estan en las dos submuestras) en
            # los que las dos soluciones coinciden
            validos = np.sum((v1 * v2) > 0)
            scores[k - 2, i] = np.sum((m1 * m2)[np.triu_indices(n, k=1)] > 0) / (validos * (validos - 1) / 2)

    return scores
```

```python
# funciones para graficar los resultados de cada método
def graficar_gap(gap, s, titulo):
    plt.figure(figsize=(6, 4))
    plt.errorbar(range(1, len(gap) + 1), gap, yerr=s, marker="o", capsize=3)
    plt.xlabel("k")
    plt.ylabel("gap")
    plt.title("Gap Statistic - " + titulo)
    plt.grid(alpha=0.3)
    plt.show()

# distribución acumulada de los scores para cada k
def graficar_estabilidad(scores, titulo):
    plt.figure(figsize=(6, 4))
    for i in range(scores.shape[0]):
        plt.plot(np.sort(scores[i]), np.linspace(0, 1, scores.shape[1]), label=f"k={i + 2}")
    plt.xlabel("score de estabilidad")
    plt.ylabel("proporción acumulada")
    plt.title("Estabilidad - " + titulo)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()
```

### 4 Gausianas

```python
np.random.seed(0)
tot_puntos = 100
sep = 2
medias = [[-sep, -sep], [2 * sep, 0], [0.7 * sep, 2.5 * sep], [-sep, sep]]
X_gaus = np.vstack([np.random.normal(mu, 0.5, size=(tot_puntos, 2)) for mu in medias])
clases_gaus = np.repeat(np.arange(4), tot_puntos)
    
plt.figure(figsize=(6, 5))
plt.scatter(X_gaus[:, 0], X_gaus[:, 1], c=clases_gaus, cmap="tab10", s=20)
plt.title("4 Gausianas")
plt.grid(alpha=0.3)
plt.show()
```

```python
np.random.seed(0)
k_gap, gap, s = gap_statistic(X_gaus)
print("Gap Statistic elige k =", k_gap)
graficar_gap(gap, s, "4 Gausianas")
```

```python
np.random.seed(0)
scores = estabilidad(X_gaus, m=50, f=0.8)
print(pd.DataFrame({"k": range(2, 11), "score medio": scores.mean(axis=1),
                    "percentil 10": np.percentile(scores, 10, axis=1)}).round(3).to_string(index=False))
graficar_estabilidad(scores, "4 Gausianas")
```

Los dos métodos encuentran los 4 clusters.

Gap Statistic elige directamente k=4, y en el grafico, el Gap(4) es claramente mayor al Gap(5) menos su barra de error. Esto puede verse muy claramente debido a que el problema de las 4 Gausianas con baja dispersión es muy facil de resolver para métodos de clustering.

En estabilidad, para k=2, 3 y 4 todas las repeticiones dan un score de 1, es decir que las dos soluciones coinciden siempre, y a partir de k=5 la distribución se corre hacia la izquierda (score medio de 0.971 y percentil 10 de 0.935). Se elige el mayor k antes de que se pierda la estabilidad, que es k=4. Que k=2 y k=3 también sean estables tiene sentido: como hay gausianas más cerca entre sí que de las demás, k-means las junta siempre de la misma forma cuando se le piden menos clusters, y esas soluciones también se repiten en cada submuestra.

Gap Statistic vuelve a elegir k=4, pero las barras de error son mucho más grandes que en los otros datasets, y desde k=4 en adelante el gap es casi constante, por lo que la elección no es muy confiable.

En estabilidad, k=2 es el más estable (0.984), seguido de k=4 (0.975), mientras que k=3 y k≥5 son menos estables.

Esto se vio en el ej. 1, el hecho de que la estructura más fuerte de los datos es el año de la medición. La segunda estructura más estable es la de los 4 clusters, que podría corresponder al par (año, especie).

Gap Statistic eligió k=4 en los tres datasets. En las 4 gausianas es la respuesta correcta, en Lampone, se pueden tomar a las combinaciones de año y especie como 4 clases distintas, por lo que la solución también resulta correcta. Sin embargo, en el caso de Iris, Gap Statistic determina 4 clústers, cuando solo existen 3 clases, aunque la diferencia entre el gap de 3 y 4 clústers es muy poca. Esto puede verse claramente en los gráficos de estabilidad. Para el caso de las 4 gausianas, se ve perfectamente la solución estable de 4 clústers, pero en el caso de Iris, la solución estable puede variar entre los 2 y 3 clústers.

Se puede notar con este análisis que estabilidad busca la solución más robusta, por lo que busca las estructuras más fuertes. Mientras que Gap Statistic busca el k que más separa los datos.

En general, resulta más conveniente analizar los gráficos y tener un entendimiento de los datos para determinar el mejor k a utilizar, que solo basarse en la respuesta por ejemplo del k que otorga Gap Statistic.

# Ejercicio 4

Se usa el dataset "Human Activity Recognition with Smartphones". Tiene 2947 mediciones de los sensores del celular (acelerómetro y giroscopio) de personas mientras realizan 6 actividades distintas: caminar, subir escaleras, bajar escaleras, estar sentado, parado o acostado. Cada medición tiene 561 variables.

La idea es analizar los datos sin mirar las clases, ver cuántos clusters sugiere Gap Statistic, agrupar con k-means, y al final comparar los clusters con las actividades reales.

```python
activity = pd.read_csv("../tp2/activity.csv")
X_act = scale(activity.drop(columns=["subject", "Activity"]).to_numpy())
```

```python
# PCA para ver los datos proyectados sin mirar las clases
pca_act = PCA()
X_act_pca = pca_act.fit_transform(X_act)
var_act = pca_act.explained_variance_ratio_

plt.figure(figsize=(7, 5))
plt.scatter(X_act_pca[:, 0], X_act_pca[:, 1], s=8, alpha=0.4)
plt.xlabel("PC1 (%.1f%%)" % (var_act[0] * 100))
plt.ylabel("PC2 (%.1f%%)" % (var_act[1] * 100))
plt.title("PCA Activity")
plt.grid(alpha=0.3)
plt.show()
```

Como las 561 variables no se pueden graficar, se proyectan los datos con PCA. Las dos primeras componentes explican el 56.8% de la varianza, y para llegar al 90% hacen falta 61 componentes, lo que muestra que hay mucha redundancia en las variables. Aún así, en la proyección se ven claramente dos grupos separados, por lo que se esperaría encontrar al menos dos clústers.

```python
np.random.seed(0)
k_gap, gap, s = gap_statistic(X_act, k_max=30)
graficar_gap(gap, s, "Activity")
```

En este caso la regla de Gap Statistic no sirve para elegir k: el gap crece para todos los k, por lo que el método elige el máximo que se probó (k=10). Esto pasa porque con casi 3000 puntos el desvío s es muy chico (0.002), y cualquier aumento mínimo del gap alcanza para que la regla no se cumpla. Se probó también con k hasta 15 y el resultado es el mismo.

Sin embargo, en el gráfico se ve claramente un codo en k=2: el gap aumenta 0.461 de k=1 a k=2, después 0.083 de k=2 a k=3, y a partir de ahí los aumentos son menores a 0.02. Es decir que la ganancia más importante es pasar de 1 a 2 clusters, y los clusters que se agregan después mejoran muy poco la solución. Esto coincide con los dos grupos que se veían en la proyección de PCA, por lo que se elige k=2.

```python
# k-means con k=2, mostrando los clusters sobre la proyección de PCA
clusters_act = KMeans(n_clusters=2, n_init=10, random_state=0).fit(X_act).labels_
print("tamaño de los clusters:", np.bincount(clusters_act))

plt.figure(figsize=(7, 5))
plt.scatter(X_act_pca[:, 0], X_act_pca[:, 1], c=clusters_act, cmap="tab10", s=8, alpha=0.4)
plt.xlabel("PC1 (%.1f%%)" % (var_act[0] * 100))
plt.ylabel("PC2 (%.1f%%)" % (var_act[1] * 100))
plt.title("k-means con k=2")
plt.grid(alpha=0.3)
plt.show()
```

K-means con k=2 separa exactamente los dos grupos que se veían en la proyección de PCA, con 1390 y 1557 puntos.

```python
# recién ahora miro las actividades reales, para ver a qué corresponde cada cluster
print(pd.crosstab(activity["Activity"], clusters_act, colnames=["cluster"]))

# y pruebo con k=6, la cantidad de actividades que hay en el dataset
clusters_act6 = KMeans(n_clusters=6, n_init=10, random_state=0).fit(X_act).labels_
print()
print(pd.crosstab(activity["Activity"], clusters_act6, colnames=["cluster"]))
```

Al comparar con las actividades reales, los dos clusters corresponden a las actividades dinámicas (caminar, subir y bajar escaleras) y a las estáticas (sentado, parado y acostado). Solo 3 de los 2947 puntos quedan en el cluster equivocado. Esto coincide con lo visto en el TP2, donde PC1 separaba perfectamente estos dos tipos de actividades.

Con k=6, que es la cantidad real de actividades, k-means no logra separarlas. Acostado queda bastante bien separado (446 de 537 en un mismo cluster), pero sentado y parado se reparten entre los mismos dos clusters, y las tres formas de caminar quedan mezcladas en otros tres. Ningún cluster de las actividades dinámicas mezcla con las estáticas, por lo que la separación principal se mantiene.

### Conclusión

La estructura más fuerte de los datos es la división entre actividades estáticas y dinámicas. Se ve en la proyección de PCA sin mirar las clases, Gap Statistic muestra su mayor salto en k=2, y k-means con 2 clusters la recupera casi perfectamente.

Las 6 actividades, en cambio, no forman clusters naturales. Las diferencias entre caminar, subir o bajar escaleras, o entre estar sentado o parado, son mucho más chicas que la diferencia entre moverse y estar quieto, y además quedan repartidas entre muchas variables (hacen falta 61 componentes para explicar el 90% de la varianza). Por eso k-means con k=6 mezcla actividades, y Gap Statistic no encuentra un k claro más allá de 2. Para distinguir esas actividades hace falta usar las clases, como en el TP2, donde con métodos supervisados se lograba separarlas.

También se vio una limitación de Gap Statistic: con muchos datos el desvío s es tan chico que la regla deja de ser útil, y hay que interpretar la forma de la curva en lugar de quedarse con el k que devuelve el método.
